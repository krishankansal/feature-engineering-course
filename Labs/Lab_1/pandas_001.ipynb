### Understanding a DataFrame
### Creating data
There are two core objects in pandas: the DataFrame and the Series.

#### DataFrame
A DataFrame is a table. It contains an array of individual entries, each of which has a certain value. 
Each entry corresponds to a row (or record) and a column.

For example, consider the following simple DataFrame:

```python
df = pd.DataFrame({'Yes': [50, 21], 'No': [131, 2]})
```

```python
df
```

```python
import pandas as pd
students = {
    "Roll_No": [101, 102, 103, 104, 105,106],
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Vikas","Samar"],    
    "Department": ["BCA", "BCA", "MCA", "BCA", "MCA","BCA"],   
    "Marks": [82, 91, 76, 88, 69,92],
    "Attendance": [92, 95, 87, 90, 85,99]
}

df = pd.DataFrame(students)
df
```

```python
# Shows the first 5 rows of your dataset to preview it.
df.head()
```

```python
# Gives total dimensions as a tuple (number of rows, number of columns).
df.shape
```

```python
# Lists the names of all columns in the dataset.
df.columns
```

```python
# Calculates the average value
df["Marks"].mean()  
```

```python
# Lists the data type (like integers, floats, or text) of every column.
df.dtypes
```
