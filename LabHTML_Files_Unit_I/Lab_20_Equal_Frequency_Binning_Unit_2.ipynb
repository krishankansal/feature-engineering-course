# 2. Equal-Frequency Binning

Before writing `pd.qcut()`, let's first understand **what problem Equal-Frequency Binning is trying to solve**.


## 2.1 Why do we need Equal-Frequency Binning?

Recall our marks:

```text
12, 18, 45, 67, 71, 72, 78, 82, 85, 88, 91, 95

With **Equal-Width Binning**, we divided the numerical range into three equal-width intervals:


12 -------- 39.67 -------- 67.33 -------- 95
      Low          Medium          High

The result was:

Low       → 2 students
Medium    → 2 students
High      → 8 students
```
So we have a problem:

#### The bins have equal width, but the observations are very unevenly distributed.

This happens because the data itself is not uniformly distributed.

## 2.2 The Basic Idea

What if, instead of asking:

> **"Can I divide the numerical range into equal-sized intervals?"**

we ask:

> **"Can I divide the observations into groups containing approximately the same number of observations?"**

That is the idea behind **Equal-Frequency Binning**.

### Definition

> **Equal-Frequency Binning divides the observations into bins containing approximately the same number of observations.**

Here, the important word is **frequency**.

**Frequency = Number of observations**

## 2.3 Simple Example

Suppose we have 12 marks:


Suppose we want 3 bins.

There are:

**12 observations**

So ideally:

**12/3 = 4**

We want approximately 4 observations in each bin.


### Conceptually

| Bin 1            | Bin 2            | Bin 3            |
|------------------|------------------|------------------|
| 12, 18, 45, 67   | 71, 72, 78, 82   | 85, 88, 91, 95   |
| ←--- 4 obs ---→  | ←--- 4 obs ---→  | ←--- 4 obs ---→  |


### Observed Intervals

- **Bin 1:** 12–67  
- **Bin 2:** 71–82  
- **Bin 3:** 85–95  

Notice something interesting:

- The widths are **not equal**.  
- But the number of observations is **approximately equal**.  

That is the **fundamental difference** between Equal-Width and Equal-Frequency binning.


## 2.4 Equal-Width vs Equal-Frequency



| Feature                | Equal-Width              | Equal-Frequency              |
|-------------------------|--------------------------|------------------------------|
| **Main objective**      | Equal interval width     | Equal number of observations |
| **Focus**               | Range                    | Frequency                    |
| **Bin width**           | Equal                    | May be different             |
| **Observations per bin**| May be different         | Approximately equal          |
| **Pandas**              | `pd.cut()`               | `pd.qcut()`                  |


**Remember**

- **Equal-Width → Equal size of intervals**  
- **Equal-Frequency → Equal number of observations**

This distinction should be very clear before we introduce `qcut()`.


## 2.5 How does Equal-Frequency know where to divide?

This is where **quantiles** enter the picture.

Suppose we want **4 equal-frequency bins**.

We want approximately:
```text

25%       25%       25%       25%
 |---------|---------|---------|
            Q1       Q2       Q3
```

### Boundaries are determined using:

- **Q1 → 25th percentile**  
- **Q2 → 50th percentile (Median)**  
- **Q3 → 75th percentile**


### Visual Representation
```text

                    Q1       Q2       Q3
                     ↓        ↓        ↓
25%                  50%      75%
 |--------------------|--------|--------|
        25%              25%      25%       25%
```

### Note


**Quantiles help us find the points that divide ordered data into approximately equal portions.**




## 2.6 Q1, Q2 and Q3 — quick revision

We already discussed these, so now connect them directly to binning.

| Quantile | Percentage | Meaning |
|----------|------------|---------|
| **Q1**   | 25%        | 25% of observations are approximately below this point |
| **Q2**   | 50%        | Middle of the data; Median |
| **Q3**   | 75%        | 75% of observations are approximately below this point |


### Therefore:

- **Q1 = 25%**  
- **Q2 = 50%**  
- **Q3 = 75%**


These become useful boundaries when creating **four equal-frequency bins**.


````python
import pandas as pd

marks = [12, 18, 45, 67, 71, 72, 78, 82, 85, 88, 91, 95]

df = pd.DataFrame({"marks": marks})

df

````

````python
df["grade_equalfrequency"] = pd.qcut(
    df["marks"],
    q=3,
    labels=["Low", "Medium", "High"]
)
df
````

One more useful experiment
Now let's compare both methods in the same DataFrame.


````python
df["grade_equalwidth"] = pd.cut(
    df["marks"],
    bins=3,
    labels=["Low", "Medium", "High"]
)

df["grade_equalfrequency"] = pd.qcut(
    df["marks"],
    q=3,
    labels=["Low", "Medium", "High"]
)

df
````
