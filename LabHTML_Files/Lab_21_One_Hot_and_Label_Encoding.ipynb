#### Label Encoding 

````python
from sklearn.preprocessing import LabelEncoder
# Let's create our data:
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Department": ["BCA", "MCA", "BBA", "BCA", "MCA"]
})

print(df)
# Now create the encoder:
encoder = LabelEncoder()
# Fit and transform:
df["Department_encoded"] = encoder.fit_transform(df["Department"])
# Let's see the result:
print(df)
````

````python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    'Education': ['Graduate', 'PhD', 'School', 'Postgraduate', 'Graduate']
})

encoder = OrdinalEncoder(
    categories=[['School', 'Graduate', 'Postgraduate', 'PhD']]
)

df['Education_Encoded'] = encoder.fit_transform(df[['Education']])

print(df)
````

````python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# --------------------------------------------------
# 1. Create the dataset
# --------------------------------------------------

df = pd.DataFrame({
    "Student": ["A", "B", "C"],
    "Satisfaction": ["Low", "High", "Medium"]
})

print("Original Data:")
print(df)


# --------------------------------------------------
# 2. Define the meaningful order
# --------------------------------------------------

order = [["Low", "Medium", "High"]]


# --------------------------------------------------
# 3. Create the OrdinalEncoder
# --------------------------------------------------

encoder = OrdinalEncoder(categories=order)


# --------------------------------------------------
# 4. Encode the categorical feature
# --------------------------------------------------

df["Satisfaction_Encoded"] = encoder.fit_transform(
    df[["Satisfaction"]]
).astype(int)


# --------------------------------------------------
# 5. Display the encoded data
# --------------------------------------------------

print("\nEncoded Data:")
print(df)


# --------------------------------------------------
# 6. Display the mapping
# --------------------------------------------------

print("\nCategory Mapping:")
for category, value in zip(order[0], range(len(order[0]))):
    print(f"{category} → {value}")
````
